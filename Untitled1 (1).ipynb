```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x='Rating', data=df)
plt.title('Rating Distribution')
plt.xlabel('Rating (1-5)')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(x='Department Name', data=df, order=df['Department Name'].value_counts().index)
plt.title('Reviews by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(df['Age'], bins=30, kde=True)
plt.title('Age Distribution of Reviewers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
from textblob import TextBlob

def get_sentiment(text):
    return TextBlob(str(text)).sentiment.polarity

df['sentiment'] = df['Review Text'].apply(get_sentiment)
df['sentiment_label'] = df['sentiment'].apply(lambda x: 'Positive' if x > 0 else ('Negative' if x < 0 else 'Neutral'))

df['sentiment_label'].value_counts()
```

```python
sns.countplot(x='sentiment_label', data=df, order=['Positive', 'Neutral', 'Negative'])
plt.title('Sentiment Analysis of Reviews')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()
```
